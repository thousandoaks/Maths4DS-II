# Week 1 — Pooled Cross Sections: A Healthcare Example

**Maths for Data Science II**

**Pooled cross sections** stack two or more cross sections, each drawn **independently** at a
different point in time.

Here: a survey of **8 patients in 2024** and a *different* survey of **8 patients in 2026**,
combined into one dataset of 16 rows. Note the patient numbers **do not repeat** across years —
these are different people, so nobody can be followed over time.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(2024)
```

## The dataset

```python
n = 8

# 2024 survey: patients 1-8
s24 = pd.DataFrame({
    "patient_id": np.arange(1, n + 1),
    "year": 2024,
    "sbp": np.round(rng.normal(150, 8, n), 1),
})

# 2026 survey: eight DIFFERENT patients, 101-108
s26 = pd.DataFrame({
    "patient_id": np.arange(101, 100 + n + 1),
    "year": 2026,
    "sbp": np.round(rng.normal(142, 8, n), 1),
})

pooled = pd.concat([s24, s26], ignore_index=True)

print("Shape:", pooled.shape)
pooled
```

## The two samples side by side

```python
plt.figure(figsize=(7, 5))

for year, colour in [(2024, "tab:blue"), (2026, "tab:orange")]:
    d = pooled[pooled.year == year]
    jitter = rng.normal(0, 0.03, len(d))          # spread points out so they don't overlap
    plt.scatter(d["year"] + jitter, d["sbp"], color=colour, s=60, label=f"{year} sample")
    plt.hlines(d["sbp"].mean(), year - 0.15, year + 0.15, color=colour, lw=2)

plt.xticks([2024, 2026])
plt.xlabel("Survey year")
plt.ylabel("Systolic blood pressure (mmHg)")
plt.title("Two independent samples of patients, two years apart")
plt.legend()
plt.tight_layout()
plt.show()
```

Each dot is a different patient; the short horizontal line is that year's average.
We can compare the **2024 population** with the **2026 population**, but there are no lines
joining the dots — no patient appears in both years. That is what separates pooled cross
sections from panel data.
